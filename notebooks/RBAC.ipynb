```python
| Commande       | Sert à                                    |
| -------------- | ----------------------------------------- |
| `\l`           | Lister les **bases de données**           |
| `\l+`          | Bases + **détails** (taille, privilèges…) |
| `\c nom_db`    | Se **connecter** à une base               |
| `\conninfo`    | Voir la **connexion actuelle**            |
| `\dn`          | Lister les **schémas**                    |
| `\dn+`         | Schémas + **privilèges**                  |
| `\dt`          | Lister les **tables**                     |
| `\dt schema.*` | Tables d'un schéma précis                 |
| `\dt+`         | Tables + **taille**                       |
| `\dv`          | Lister les **views**                      |
| `\dv+`         | Views + détails                           |
| `\d table`     | Décrire une **table**                     |
| `\d+ table`    | Table + détails supplémentaires           |
| `\di`          | Lister les **indexes**                    |
| `\du`          | Lister les **rôles/utilisateurs**         |
| `\du+`         | Rôles + détails                           |
| `\dp`          | Voir les **privilèges des tables/views**  |
| `\dn+`         | Voir les **privilèges des schémas**       |
| `\df`          | Lister les **fonctions**                  |
| `\dx`          | Extensions installées                     |
| `\q`           | Quitter `psql`                            |
```

```python
\l       → PostgreSQL
\dn+     → schemas
\dt      → tables
\d       → structure
\di      → indexes
\du      → users/roles
\dp      → permissions
```
