Étape suivante — quand créer un index ?

Retenons d'abord une règle simple :

On crée un index sur une colonne souvent utilisée pour filtrer, joindre ou trier, surtout si la recherche est sélective.


````python
On a couvert
EXPLAIN
Seq Scan
Index Scan
statistiques et estimations
sélectivité
index composite
EXPLAIN ANALYZE
actual rows
Rows Removed by Filter
Buffers
Planning Time
Execution Time
````

Pourquoi B-tree est le choix par défaut ?

Parce qu'il couvre énormément de besoins courants : égalité, comparaisons et tri.

Les autres types d'index ont des usages plus spécifiques :

B-tree → recherches classiques, comparaisons, ORDER BY
Hash → égalité exacte
GIN → tableaux, JSONB, recherche de plusieurs éléments
GiST → données géométriques, recherche spécialisée
BRIN → très grandes tables où les données suivent naturellement l'ordre physique

Pour notre projet Olist, B-tree sera de loin le type que tu rencontreras le plus souvent.
